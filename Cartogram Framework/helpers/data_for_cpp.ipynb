```python
# Change these if your column names differ
STATE_COLUMN = r"State\UT"      # e.g. "State", "NAME_1", etc.
YEAR_COLUMN = "2015-16"
INPUT = "/home/kuenem/Documents/development/lectures/Master Thesis/Cartogram Framework/data/statistics/india/State_wise_SDP_01_08_2023_Rev.xls"
OUTPUT = "/home/kuenem/Documents/development/lectures/Master Thesis/Cartogram Framework/data/statistics/india/State_wise_SDP_01_08_2023_Rev.csv"
```

```python
# parser = argparse.ArgumentParser(
#     description="Convert SDP Excel file to cartogram CSV."
# )
# parser.add_argument("input", help="Input .xls file")
# parser.add_argument("year", help="Year column to extract")
# parser.add_argument("output", help="Output CSV filename")

# args = parser.parse_args()

# Requires: pip install pandas xlrd
df = pd.read_excel(INPUT, header=4)

print(df.head())
```

```python
if STATE_COLUMN not in df.columns:
    raise ValueError(f"'{STATE_COLUMN}' column not found.")

if YEAR_COLUMN not in df.columns:
    raise ValueError(
        f"Year '{YEAR_COLUMN}' not found.\n"
        f"Available columns:\n{list(df.columns)}"
    )

out = (
    df[[STATE_COLUMN, YEAR_COLUMN]]
    .rename(columns={
        STATE_COLUMN: "NAME_1",
        YEAR_COLUMN: "Data"
    })
)

# Optional: leave blank if you don't have colours yet
out["Color"] = ""

out = out.dropna(subset=["NAME_1"])

# Remove the "(2)", "(3)", etc. row
out = out[~out["NAME_1"].astype(str).str.startswith("(")]
out = out[~out["NAME_1"].astype(str).str.startswith("Source")]
out = out[~out["NAME_1"].astype(str).str.startswith('"*')]

out.to_csv("output.csv", index=False)

out.to_csv(OUTPUT, index=False)

print(f"Wrote {len(out)} rows to {OUTPUT}")
```
