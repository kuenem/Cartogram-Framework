```python
df = pd.read_csv(
    "/home/kuenem/Documents/development/lectures/Master Thesis/data/statistics/Germany/counties_de.csv",
    skiprows=4,
    sep=";",
    header=None
)

print(df.shape)
print(df.columns.tolist())
print(df.head(3))
```

```python
rows = []

# Row 1 contains the dates
year_row = df.iloc[1]

for _, row in df.iloc[2:].iterrows():

    code = str(row[0]).zfill(5)
    entity = str(row[1]).split(",")[0].strip()

    for col in [2, 4, 6, 8, 10]:  # population columns only

        year = int(str(year_row[col]).split(".")[-1])

        rows.append({
            "Entity": entity,
            "Code": code,
            "Year": year,
            "Population": pd.to_numeric(row[col], errors="coerce")
        })

result = pd.DataFrame(rows)

result = result.sort_values(
    ["Entity", "Year"]
).reset_index(drop=True)
```
