```python
initial_amount = 0
monthly_investment = 1500
yearly_investment = 12 * monthly_investment
annunal_growth_rate = 0.08
years = 6

growth = 0
growth_growth = 0

for year in range(years):
    initial_amount += yearly_investment
    growth_growth += growth * annunal_growth_rate + growth_growth * annunal_growth_rate
    growth = initial_amount * annunal_growth_rate
    total_amount = initial_amount + growth + growth_growth
    print(f"Year {year + 1}: Total Amount: {total_amount:.2f}, Growth: {growth:.2f}, Growth of Growth: {growth_growth:.2f}")
```
